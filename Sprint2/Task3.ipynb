Описание задачи:

Создать базовый класс Inventory в C#, который будет представлять информацию о наличии товаров на складе. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Inventory:

• Атрибуты: ID склада (WarehouseId), Название склада (WarehouseName), Общий объем хранения (StorageCapacity).

• Методы:

  ○ GetStorageStatus(): метод для получения статуса доступного пространства на складе.

  ○ AddItem(Item item): метод для добавления товара на склад.

  ○ RemoveItem(Item item): метод для удаления товара со склада.

Требования к производным классам:

1. ПерсональныйСклад (PersonalInventory): Должен содержать дополнительные атрибуты, такие как Владелец склада (OwnerName). Метод GetStorageStatus() должен быть переопределен для отображения информации о владельце склада вместе с статусом хранения.

2. ГрупповойСклад (GroupInventory): Должен содержать дополнительные атрибуты, такие как Группа товаров (ProductGroup). Метод AddItem() должен быть переопределен для добавления информации о группе товаров при добавлении нового товара.

3. АвтоматизированныйСклад (AutomatedInventory) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Автоматизация уровня (AutomationLevel). Метод RemoveItem() должен быть переопределен для добавления информации о уровне автоматизации при удалении товара.

#### Дополнительное задание к прошлому таску :
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

#### Дополнительное задание :
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
// Интерфейсы для множественного наследования

interface IReportable
{
    void CreateReport();
}

interface IReservable
{
    void ReservePlace();
}


// Класс товара

class Item
{
    private string name;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public Item(string name)
    {
        Name = name;
    }
}


// Базовый класс Inventory

class Inventory
{
    private int warehouseId;
    private string warehouseName;
    private double storageCapacity;

    public int WarehouseId
    {
        get { return warehouseId; }
        set { warehouseId = value; }
    }

    public string WarehouseName
    {
        get { return warehouseName; }
        set { warehouseName = value; }
    }

    public double StorageCapacity
    {
        get { return storageCapacity; }
        set { storageCapacity = value; }
    }


    // Новые атрибуты Sprint2

    public string Address { get; set; }
    public int CurrentItemCount { get; set; }
    public bool IsOpen { get; set; }


    public Inventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity)
    {
        WarehouseId = warehouseId;
        WarehouseName = warehouseName;
        StorageCapacity = storageCapacity;
    }


    public virtual void GetStorageStatus()
    {
        Console.WriteLine($"ID склада: {WarehouseId}");
        Console.WriteLine($"Название склада: {WarehouseName}");
        Console.WriteLine($"Объем хранения: {StorageCapacity}");
    }

    public virtual void AddItem(Item item)
    {
        Console.WriteLine(
            $"Товар {item.Name} добавлен на склад {WarehouseName}"
        );
    }

    public virtual void RemoveItem(Item item)
    {
        Console.WriteLine(
            $"Товар {item.Name} удален со склада {WarehouseName}"
        );
    }


    // Новые методы Sprint2

    public void ShowAddress()
    {
        Console.WriteLine($"Адрес склада: {Address}");
    }

    public void ShowItemCount()
    {
        Console.WriteLine($"Количество товаров: {CurrentItemCount}");
    }

    public void OpenWarehouse()
    {
        IsOpen = true;
        Console.WriteLine($"Склад {WarehouseName} открыт");
    }
}


// Персональный склад
// Простое наследование: PersonalInventory -> Inventory

class PersonalInventory : Inventory
{
    private string ownerName;

    public string OwnerName
    {
        get { return ownerName; }
        set { ownerName = value; }
    }


    // Новые атрибуты Sprint2

    public string AccessCode { get; set; }
    public int RoomNumber { get; set; }
    public bool IsLocked { get; set; }


    public PersonalInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string ownerName)
        : base(warehouseId, warehouseName, storageCapacity)
    {
        OwnerName = ownerName;
    }


    public override void GetStorageStatus()
    {
        base.GetStorageStatus();
        Console.WriteLine($"Владелец склада: {OwnerName}");
    }


    // Новые методы Sprint2

    public void ShowOwnerInfo()
    {
        Console.WriteLine($"Владелец: {OwnerName}");
        Console.WriteLine($"Номер помещения: {RoomNumber}");
    }

    public void LockStorage()
    {
        IsLocked = true;
        Console.WriteLine("Персональный склад закрыт");
    }

    public void UnlockStorage()
    {
        IsLocked = false;
        Console.WriteLine("Персональный склад открыт");
    }
}


// Групповой склад
// Множественное наследование через интерфейсы

class GroupInventory : Inventory, IReportable, IReservable
{
    private string productGroup;

    public string ProductGroup
    {
        get { return productGroup; }
        set { productGroup = value; }
    }


    // Новые атрибуты Sprint2

    public int EmployeeCount { get; set; }
    public string SectionName { get; set; }
    public string ResponsiblePerson { get; set; }


    public GroupInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string productGroup)
        : base(warehouseId, warehouseName, storageCapacity)
    {
        ProductGroup = productGroup;
    }


    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Группа товаров: {ProductGroup}");
    }


    // Новые методы Sprint2

    public void ShowGroupInfo()
    {
        Console.WriteLine($"Отдел: {SectionName}");
        Console.WriteLine($"Ответственный: {ResponsiblePerson}");
        Console.WriteLine($"Количество сотрудников: {EmployeeCount}");
    }

    public void CreateReport()
    {
        Console.WriteLine(
            $"Создан отчет по складу {WarehouseName}"
        );
    }

    public void ReservePlace()
    {
        Console.WriteLine(
            $"На складе {WarehouseName} зарезервировано место"
        );
    }
}


// Автоматизированный склад

class AutomatedInventory : Inventory
{
    private string automationLevel;

    public string AutomationLevel
    {
        get { return automationLevel; }
        set { automationLevel = value; }
    }


    // Новые атрибуты Sprint2

    public int RobotCount { get; set; }
    public string SystemName { get; set; }
    public bool IsSystemActive { get; set; }


    public AutomatedInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string automationLevel)
        : base(warehouseId, warehouseName, storageCapacity)
    {
        AutomationLevel = automationLevel;
    }


    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine(
            $"Уровень автоматизации: {AutomationLevel}"
        );
    }


    // Новые методы Sprint2

    public void StartSystem()
    {
        IsSystemActive = true;
        Console.WriteLine($"Система {SystemName} запущена");
    }

    public void StopSystem()
    {
        IsSystemActive = false;
        Console.WriteLine($"Система {SystemName} остановлена");
    }

    public void ShowAutomationInfo()
    {
        Console.WriteLine($"Система: {SystemName}");
        Console.WriteLine($"Количество роботов: {RobotCount}");
        Console.WriteLine($"Уровень автоматизации: {AutomationLevel}");
    }
}


// Сложное (многоуровневое) наследование:
//
// Inventory
//     ↓
// AutomatedInventory
//     ↓
// SmartAutomatedInventory

class SmartAutomatedInventory : AutomatedInventory
{
    public string ControlSystem { get; set; }

    public SmartAutomatedInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string automationLevel,
        string controlSystem)
        : base(
            warehouseId,
            warehouseName,
            storageCapacity,
            automationLevel)
    {
        ControlSystem = controlSystem;
    }

    public void ShowControlSystem()
    {
        Console.WriteLine(
            $"Система управления: {ControlSystem}"
        );
    }
}


// -------------------------------------------------
// Создание товаров
// -------------------------------------------------

Item item1 = new Item("Ноутбук");
Item item2 = new Item("Телефон");
Item item3 = new Item("Принтер");


// -------------------------------------------------
// Создание складов
// -------------------------------------------------

PersonalInventory personalInventory =
    new PersonalInventory(
        1,
        "Личный склад",
        100,
        "Юлия"
    );

personalInventory.Address = "Тюмень";
personalInventory.CurrentItemCount = 10;
personalInventory.IsOpen = true;

personalInventory.AccessCode = "1234";
personalInventory.RoomNumber = 15;
personalInventory.IsLocked = false;


GroupInventory groupInventory =
    new GroupInventory(
        2,
        "Склад электроники",
        500,
        "Электроника"
    );

groupInventory.Address = "Тюмень, ул. Республики";
groupInventory.CurrentItemCount = 250;
groupInventory.IsOpen = true;

groupInventory.EmployeeCount = 8;
groupInventory.SectionName = "Отдел электроники";
groupInventory.ResponsiblePerson = "Иван Петров";


AutomatedInventory automatedInventory =
    new AutomatedInventory(
        3,
        "Автоматизированный склад",
        1000,
        "Высокий"
    );

automatedInventory.Address = "Тюмень, Промышленная зона";
automatedInventory.CurrentItemCount = 600;
automatedInventory.IsOpen = true;

automatedInventory.RobotCount = 20;
automatedInventory.SystemName = "AutoStorage";
automatedInventory.IsSystemActive = false;


SmartAutomatedInventory smartInventory =
    new SmartAutomatedInventory(
        4,
        "Умный склад",
        1500,
        "Полный",
        "SmartControl"
    );


// -------------------------------------------------
// Проверка персонального склада
// -------------------------------------------------

Console.WriteLine("=== Персональный склад ===");

personalInventory.GetStorageStatus();
personalInventory.AddItem(item1);
personalInventory.ShowAddress();
personalInventory.ShowItemCount();
personalInventory.ShowOwnerInfo();
personalInventory.LockStorage();
personalInventory.UnlockStorage();


Console.WriteLine();


// -------------------------------------------------
// Проверка группового склада
// -------------------------------------------------

Console.WriteLine("=== Групповой склад ===");

groupInventory.GetStorageStatus();
groupInventory.AddItem(item2);
groupInventory.ShowGroupInfo();
groupInventory.CreateReport();
groupInventory.ReservePlace();


Console.WriteLine();


// -------------------------------------------------
// Проверка автоматизированного склада
// -------------------------------------------------

Console.WriteLine("=== Автоматизированный склад ===");

automatedInventory.GetStorageStatus();
automatedInventory.AddItem(item3);
automatedInventory.RemoveItem(item3);
automatedInventory.StartSystem();
automatedInventory.ShowAutomationInfo();
automatedInventory.StopSystem();


Console.WriteLine();


// -------------------------------------------------
// Сложное наследование
// -------------------------------------------------

Console.WriteLine("=== Сложное наследование ===");

smartInventory.GetStorageStatus();
smartInventory.StartSystem();
smartInventory.ShowControlSystem();
```
