### Вариант задания №14 


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Supplier в C#, который будет представлять информацию о поставщиках товаров или услуг. На основе этого класса разработать 2–3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Supplier:

Атрибуты:
- ID поставщика (SupplierId)
- Название компании (CompanyName)
- Тип продукции (ProductType)

Методы:
- GetCompanyInfo() — получение информации о компании.
- ProvideQuote() — предоставление котировки на товары или услуги.
- SubmitOrder() — отправка заказа поставщику.

Производные классы:

1. Manufacturer (Производитель):
- дополнительный атрибут FoundedYear — год основания;
- переопределить ProvideQuote(), добавив информацию о годе основания компании.

2. Retailer (Ритейлер):
- дополнительный атрибут StoreLocation — расположение магазина;
- переопределить SubmitOrder(), добавив информацию о расположении магазина.

3. Importer (Импортер):
- дополнительный атрибут OriginCountry — страна происхождения товара;
- переопределить GetCompanyInfo(), добавив информацию о стране происхождения товара.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
class Supplier
{
    public int SupplierId;
    public string CompanyName;
    public string ProductType;

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine($"ID поставщика: {SupplierId}");
        Console.WriteLine($"Название компании: {CompanyName}");
        Console.WriteLine($"Тип продукции: {ProductType}");
    }

    public virtual void ProvideQuote()
    {
        Console.WriteLine($"Компания {CompanyName} предоставляет котировку");
    }

    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}");
    }
}

class Manufacturer : Supplier
{
    public int FoundedYear;

    public override void ProvideQuote()
    {
        Console.WriteLine($"Компания {CompanyName} предоставляет котировку. Год основания: {FoundedYear}");
    }
}

class Retailer : Supplier
{
    public string StoreLocation;

    public override void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}. Расположение магазина: {StoreLocation}");
    }
}

class Importer : Supplier
{
    public string OriginCountry;

    public override void GetCompanyInfo()
    {
        Console.WriteLine($"ID поставщика: {SupplierId}");
        Console.WriteLine($"Название компании: {CompanyName}");
        Console.WriteLine($"Тип продукции: {ProductType}");
        Console.WriteLine($"Страна происхождения товара: {OriginCountry}");
    }
}

Manufacturer manufacturer = new Manufacturer();

manufacturer.SupplierId = 1;
manufacturer.CompanyName = "TechFactory";
manufacturer.ProductType = "Электроника";
manufacturer.FoundedYear = 2005;

manufacturer.GetCompanyInfo();
manufacturer.ProvideQuote();
manufacturer.SubmitOrder();


Console.WriteLine();


Retailer retailer = new Retailer();

retailer.SupplierId = 2;
retailer.CompanyName = "SmartStore";
retailer.ProductType = "Бытовая техника";
retailer.StoreLocation = "Тюмень";

retailer.GetCompanyInfo();
retailer.ProvideQuote();
retailer.SubmitOrder();


Console.WriteLine();


Importer importer = new Importer();

importer.SupplierId = 3;
importer.CompanyName = "GlobalImport";
importer.ProductType = "Автозапчасти";
importer.OriginCountry = "Германия";

importer.GetCompanyInfo();
importer.ProvideQuote();
importer.SubmitOrder();
```
