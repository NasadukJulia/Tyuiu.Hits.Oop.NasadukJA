# Спринт ревью №2
## Тема: Наследование, абстрактный класс и интерфейсы

**Цель работы:** Создать классы, представляющие структуру коммерческого предприятия, используя простое, сложное и множественное наследование через интерфейсы.

### В ходе работы:
1. Создан базовый класс `Employee` (Сотрудник).
2. Создан класс `Manager`, наследующийся от `Employee`.
3. Создан класс `ProjectManager`, наследующийся от `Manager`.
4. Созданы интерфейсы `IReportable` и `ITeamLeader`.
5. Реализованы методы `PrintInfo()`, `GenerateReport()` и `GetTeamMembers()`.
6. Проведено тестирование всех классов и методов.

```polyglot-notebook
using System;

// Базовый класс Сотрудник
class Employee
{
    public string Name;
    public string Position;
    public double Salary;

    // Конструктор
    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    // Вывод информации о сотруднике
    public virtual void PrintInfo()
    {
        Console.WriteLine("Имя: " + Name);
        Console.WriteLine("Должность: " + Position);
        Console.WriteLine("Зарплата: " + Salary);
    }
}


// Класс Менеджер наследуется от Employee
class Manager : Employee
{
    public string Department;

    // Конструктор
    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    // Переопределяем метод PrintInfo
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine("Отдел: " + Department);
    }
}


// Интерфейс для создания отчёта
interface IReportable
{
    string GenerateReport();
}


// Интерфейс для получения членов команды
interface ITeamLeader
{
    string[] GetTeamMembers();
}


// ProjectManager наследуется от Manager
// и реализует два интерфейса
class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName;

    // Конструктор
    public ProjectManager(
        string name,
        string position,
        double salary,
        string department,
        string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    // Переопределяем PrintInfo
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine("Проект: " + ProjectName);
    }

    // Метод интерфейса IReportable
    public string GenerateReport()
    {
        return "Отчёт по проекту: " + ProjectName;
    }

    // Метод интерфейса ITeamLeader
    public string[] GetTeamMembers()
    {
        return new string[] { "Анна", "Иван", "Олег" };
    }
}


// ТЕСТИРОВАНИЕ

// Создаём обычного сотрудника
Employee employee = new Employee(
    "Алексей",
    "Разработчик",
    50000
);

// Создаём менеджера
Manager manager = new Manager(
    "Мария",
    "Менеджер",
    70000,
    "IT отдел"
);

// Создаём проектного менеджера
ProjectManager projectManager = new ProjectManager(
    "Елена",
    "Проектный менеджер",
    90000,
    "Разработка",
    "Интернет-магазин"
);


// Выводим информацию о сотруднике
Console.WriteLine("СОТРУДНИК:");
employee.PrintInfo();

Console.WriteLine();


// Выводим информацию о менеджере
Console.WriteLine("МЕНЕДЖЕР:");
manager.PrintInfo();

Console.WriteLine();


// Выводим информацию о проектном менеджере
Console.WriteLine("ПРОЕКТНЫЙ МЕНЕДЖЕР:");
projectManager.PrintInfo();

Console.WriteLine();


// Проверяем метод GenerateReport
Console.WriteLine(projectManager.GenerateReport());

Console.WriteLine();


// Проверяем метод GetTeamMembers
Console.WriteLine("Члены команды:");

string[] team = projectManager.GetTeamMembers();

foreach (string member in team)
{
    Console.WriteLine(member);
}
```
