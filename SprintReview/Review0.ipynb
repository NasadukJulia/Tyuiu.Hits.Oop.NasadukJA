## Название проекта:

### Управление растениями в саду

## Описание проекта:

---

Создать проект для управления растениями в саду с использованием основных принципов ООП в C#.

Необходимо создать базовый класс `Plant` и производные классы `Flower` и `Tree`.

В проекте должны быть реализованы наследование и полиморфизм.

## Требования:

- Базовый класс `Plant`
- Производный класс `Flower`
- Производный класс `Tree`
- Свойства растений
- Методы роста и цветения
- Переопределение метода `Bloom()`
- Создание объектов цветка и дерева
- Демонстрация полиморфизма

```polyglot-notebook
using System;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"{Name} выросло. Высота: {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wilt()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowYears(int years)
    {
        Age += years;
        Console.WriteLine($"Возраст дерева {Name}: {Age} лет.");
    }
}

Flower flower = new Flower("Роза", 30, "Красный", 25);
Tree tree = new Tree("Дуб", 250, 5, "Лиственные");

flower.Bloom();
flower.Grow(5);
flower.Wilt();

Console.WriteLine();

tree.Bloom();
tree.Grow(20);
tree.GrowYears(2);

Console.WriteLine();

Plant plant1 = flower;
Plant plant2 = tree;

plant1.Bloom();
plant2.Bloom();
```
