
# Спринт ревью №3

## Тема: Полиморфизм и Generic классы

**Цель работы:** Создать программу для работы с компонентами компьютера, используя полиморфизм, наследование, интерфейсы и Generic методы.

## В ходе работы:

1. Создан абстрактный базовый класс `Component`.
2. Созданы классы `CPU`, `RAM`, `GPU` и `HDD`, наследующиеся от `Component`.
3. Создан интерфейс `IInstallable` с методом `Install()`.
4. Реализованы виртуальный метод `GetInfo()` и его переопределение в дочерних классах.
5. Переопределён метод `ToString()` и перегружены операторы сравнения `==` и `!=`.
6. Создан класс `Computer` для добавления, удаления и хранения компонентов.
7. Реализованы Generic методы `AddComponent<T>()` и `RemoveComponent<T>()`.
8. Создан статический класс `ComputerFactory` для создания компьютеров с использованием Generic.
9. Проведено тестирование всех основных классов и методов.


```polyglot-notebook

using System;
using System.Collections.Generic;

// Интерфейс для установки компонентов
interface IInstallable
{
    void Install();
}

// Базовый абстрактный класс
abstract class Component
{
    public string Name;
    public double Price;

    public Component(string name, double price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, цена: {Price}";
    }

    public override string ToString()
    {
        return GetInfo();
    }
}

// Процессор
class CPU : Component, IInstallable
{
    public double Frequency;

    public CPU(string name, double price, double frequency) : base(name, price)
    {
        Frequency = frequency;
    }

    public override string GetInfo()
    {
        return $"Процессор: {Name}, частота: {Frequency} ГГц, цена: {Price}";
    }

    public void Install()
    {
        Console.WriteLine($"Установлен процессор {Name}");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

// Оперативная память
class RAM : Component, IInstallable
{
    public int Memory;

    public RAM(string name, double price, int memory) : base(name, price)
    {
        Memory = memory;
    }

    public override string GetInfo()
    {
        return $"Оперативная память: {Name}, объем: {Memory} ГБ, цена: {Price}";
    }

    public void Install()
    {
        Console.WriteLine($"Установлена оперативная память {Name}");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

// Видеокарта
class GPU : Component, IInstallable
{
    public int Memory;

    public GPU(string name, double price, int memory) : base(name, price)
    {
        Memory = memory;
    }

    public override string GetInfo()
    {
        return $"Видеокарта: {Name}, память: {Memory} ГБ, цена: {Price}";
    }

    public void Install()
    {
        Console.WriteLine($"Установлена видеокарта {Name}");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

// Жесткий диск
class HDD : Component, IInstallable
{
    public int Capacity;

    public HDD(string name, double price, int capacity) : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return $"Жесткий диск: {Name}, объем: {Capacity} ГБ, цена: {Price}";
    }

    public void Install()
    {
        Console.WriteLine($"Установлен жесткий диск {Name}");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

// Класс компьютера
class Computer
{
    public List<Component> Components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        Components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        Components.Remove(component);
        Console.WriteLine($"Удален компонент: {component.Name}");
    }

    public double GetTotalPrice()
    {
        double total = 0;

        foreach (Component component in Components)
        {
            total += component.Price;
        }

        return total;
    }

    public void ShowComponents()
    {
        foreach (Component component in Components)
        {
            Console.WriteLine(component.GetInfo());
        }
    }
}

// Фабрика для создания компьютеров
static class ComputerFactory
{
    public static Computer CreateComputer<T1, T2>(T1 first, T2 second)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
    {
        Computer computer = new Computer();
        computer.AddComponent(first);
        computer.AddComponent(second);
        return computer;
    }
}

// Проверка программы
{
    CPU cpu = new CPU("Intel Core i5", 20000, 3.5);
    RAM ram = new RAM("Kingston", 5000, 16);
    GPU gpu = new GPU("NVIDIA RTX 3060", 30000, 12);
    HDD hdd = new HDD("Seagate", 6000, 1000);

    Console.WriteLine("Информация о компонентах:");
    Console.WriteLine(cpu.GetInfo());
    Console.WriteLine(ram.GetInfo());
    Console.WriteLine(gpu.GetInfo());
    Console.WriteLine(hdd.GetInfo());

    Console.WriteLine();
    Console.WriteLine("Проверка ToString:");
    Console.WriteLine(cpu.ToString());

    Console.WriteLine();
    Console.WriteLine("Сравнение процессоров:");
    CPU cpu2 = new CPU("Intel Core i5", 20000, 3.5);
    Console.WriteLine(cpu == cpu2);
    Console.WriteLine(cpu != cpu2);

    Console.WriteLine();
    Console.WriteLine("Создание компьютера:");
    Computer computer = new Computer();
    computer.AddComponent(cpu);
    computer.AddComponent(ram);
    computer.AddComponent(gpu);
    computer.AddComponent(hdd);

    Console.WriteLine();
    Console.WriteLine("Компоненты компьютера:");
    computer.ShowComponents();

    Console.WriteLine();
    Console.WriteLine($"Общая стоимость: {computer.GetTotalPrice()}");

    Console.WriteLine();
    computer.RemoveComponent(hdd);
    Console.WriteLine($"Стоимость после удаления: {computer.GetTotalPrice()}");

    Console.WriteLine();
    Console.WriteLine("Создание компьютера через фабрику:");
    Computer computer2 = ComputerFactory.CreateComputer(
        new CPU("AMD Ryzen 5", 18000, 3.6),
        new RAM("Corsair", 6000, 16)
    );
    computer2.ShowComponents();
    Console.WriteLine($"Общая стоимость: {computer2.GetTotalPrice()}");
}
```
