Описание задачи:

Создать базовый класс Inventory в C#, который будет представлять информацию о наличии товаров на складе. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Inventory:

• Атрибуты: ID склада (WarehouseId), Название склада (WarehouseName), Общий объем хранения (StorageCapacity).

• Методы:

  ○ GetStorageStatus(): метод для получения статуса доступного пространства на складе.

  ○ AddItem(Item item): метод для добавления товара на склад.

  ○ RemoveItem(Item item): метод для удаления товара со склада.

Требования к производным классам:

1. ПерсональныйСклад (PersonalInventory): Должен содержать дополнительные атрибуты, такие как Владелец склада (OwnerName). Метод GetStorageStatus() должен быть переопределен для отображения информации о владельце склада вместе с статусом хранения.

2. ГрупповойСклад (GroupInventory): Должен содержать дополнительные атрибуты, такие как Группа товаров (ProductGroup). Метод AddItem() должен быть переопределен для добавления информации о группе товаров при добавлении нового товара.

3. АвтоматизированныйСклад (AutomatedInventory) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Автоматизация уровня (AutomationLevel). Метод RemoveItem() должен быть переопределен для добавления информации о уровне автоматизации при удалении товара.

```
// -------------------------------------------------
// Интерфейсы
// -------------------------------------------------

interface IReportable
{
    void CreateReport();
}

interface IReservable
{
    void ReservePlace();
}

interface IInventoryOperations
{
    void AddItem(Item item);
    void RemoveItem(Item item);
}


// -------------------------------------------------
// Класс товара
// -------------------------------------------------

class Item
{
    private string name;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public Item(string name)
    {
        Name = name;
    }
}


// -------------------------------------------------
// Базовый класс Inventory
// -------------------------------------------------

class Inventory : IInventoryOperations
{
    private int warehouseId;
    private string warehouseName;
    private double storageCapacity;

    public int WarehouseId
    {
        get { return warehouseId; }
        set { warehouseId = value; }
    }

    public string WarehouseName
    {
        get { return warehouseName; }
        set { warehouseName = value; }
    }

    public double StorageCapacity
    {
        get { return storageCapacity; }
        set { storageCapacity = value; }
    }

    // Sprint2
    public string Address { get; set; }
    public int CurrentItemCount { get; set; }
    public bool IsOpen { get; set; }

    // Sprint3
    public string ManagerName { get; set; }
    public double Temperature { get; set; }
    public string SecurityLevel { get; set; }

    // Новые атрибуты Sprint4
    public string LastInspectionDate { get; set; }
    public double MaxWeight { get; set; }
    public bool HasFireSystem { get; set; }


    public Inventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity)
    {
        WarehouseId = warehouseId;
        WarehouseName = warehouseName;
        StorageCapacity = storageCapacity;
    }


    public virtual void GetStorageStatus()
    {
        Console.WriteLine($"ID склада: {WarehouseId}");
        Console.WriteLine($"Название склада: {WarehouseName}");
        Console.WriteLine($"Объем хранения: {StorageCapacity}");
    }

    public virtual void AddItem(Item item)
    {
        Console.WriteLine(
            $"Товар {item.Name} добавлен на склад {WarehouseName}"
        );
    }

    public virtual void RemoveItem(Item item)
    {
        Console.WriteLine(
            $"Товар {item.Name} удален со склада {WarehouseName}"
        );
    }


    // Sprint2

    public void ShowAddress()
    {
        Console.WriteLine($"Адрес склада: {Address}");
    }

    public void ShowItemCount()
    {
        Console.WriteLine($"Количество товаров: {CurrentItemCount}");
    }

    public void OpenWarehouse()
    {
        IsOpen = true;
        Console.WriteLine($"Склад {WarehouseName} открыт");
    }


    // Sprint3

    public void ShowManager()
    {
        Console.WriteLine($"Управляющий складом: {ManagerName}");
    }

    public void ChangeTemperature(double newTemperature)
    {
        Temperature = newTemperature;
        Console.WriteLine($"Температура на складе: {Temperature}");
    }

    public void ShowSecurityLevel()
    {
        Console.WriteLine($"Уровень безопасности: {SecurityLevel}");
    }

    public void GetStorageStatus(string message)
    {
        Console.WriteLine(message);
        GetStorageStatus();
    }


    // Новые методы Sprint4

    public void ShowInspectionDate()
    {
        Console.WriteLine($"Дата последней проверки: {LastInspectionDate}");
    }

    public void ShowMaxWeight()
    {
        Console.WriteLine($"Максимальный вес: {MaxWeight}");
    }

    public void CheckFireSystem()
    {
        Console.WriteLine($"Пожарная система: {HasFireSystem}");
    }
}


// -------------------------------------------------
// PersonalInventory
// -------------------------------------------------

class PersonalInventory : Inventory
{
    private string ownerName;

    public string OwnerName
    {
        get { return ownerName; }
        set { ownerName = value; }
    }

    // Sprint2
    public string AccessCode { get; set; }
    public int RoomNumber { get; set; }
    public bool IsLocked { get; set; }

    // Sprint3
    public string LastAccessDate { get; set; }
    public int MaxPersonalItems { get; set; }
    public bool HasInsurance { get; set; }

    // Новые атрибуты Sprint4
    public string StorageType { get; set; }
    public int RentalMonths { get; set; }
    public bool PaymentStatus { get; set; }


    public PersonalInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string ownerName)
        : base(warehouseId, warehouseName, storageCapacity)
    {
        OwnerName = ownerName;
    }


    public override void GetStorageStatus()
    {
        base.GetStorageStatus();
        Console.WriteLine($"Владелец склада: {OwnerName}");
    }


    // Sprint2

    public void ShowOwnerInfo()
    {
        Console.WriteLine($"Владелец: {OwnerName}");
        Console.WriteLine($"Номер помещения: {RoomNumber}");
    }

    public void LockStorage()
    {
        IsLocked = true;
        Console.WriteLine("Персональный склад закрыт");
    }

    public void UnlockStorage()
    {
        IsLocked = false;
        Console.WriteLine("Персональный склад открыт");
    }


    // Sprint3

    public void ShowLastAccess()
    {
        Console.WriteLine($"Последний доступ: {LastAccessDate}");
    }

    public void ShowPersonalLimit()
    {
        Console.WriteLine($"Лимит товаров: {MaxPersonalItems}");
    }

    public void ShowInsurance()
    {
        Console.WriteLine($"Страхование склада: {HasInsurance}");
    }


    // Новые методы Sprint4

    public void ShowStorageType()
    {
        Console.WriteLine($"Тип хранения: {StorageType}");
    }

    public void ShowRentalMonths()
    {
        Console.WriteLine($"Срок аренды: {RentalMonths} месяцев");
    }

    public void ShowPaymentStatus()
    {
        Console.WriteLine($"Оплата произведена: {PaymentStatus}");
    }
}


// -------------------------------------------------
// GroupInventory
// -------------------------------------------------

class GroupInventory : Inventory, IReportable, IReservable
{
    private string productGroup;

    public string ProductGroup
    {
        get { return productGroup; }
        set { productGroup = value; }
    }

    // Sprint2
    public int EmployeeCount { get; set; }
    public string SectionName { get; set; }
    public string ResponsiblePerson { get; set; }

    // Sprint3
    public string CategoryCode { get; set; }
    public int MaxEmployees { get; set; }
    public bool HasLoadingZone { get; set; }

    // Новые атрибуты Sprint4
    public string DepartmentCode { get; set; }
    public int ShiftCount { get; set; }
    public bool HasForklift { get; set; }


    public GroupInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string productGroup)
        : base(warehouseId, warehouseName, storageCapacity)
    {
        ProductGroup = productGroup;
    }


    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Группа товаров: {ProductGroup}");
    }


    public void ShowGroupInfo()
    {
        Console.WriteLine($"Отдел: {SectionName}");
        Console.WriteLine($"Ответственный: {ResponsiblePerson}");
        Console.WriteLine($"Количество сотрудников: {EmployeeCount}");
    }


    // Явная реализация интерфейсов Sprint4

    void IReportable.CreateReport()
    {
        Console.WriteLine(
            $"Создан отчет по складу {WarehouseName}"
        );
    }

    void IReservable.ReservePlace()
    {
        Console.WriteLine(
            $"На складе {WarehouseName} зарезервировано место"
        );
    }


    // Sprint3

    public void ShowCategoryCode()
    {
        Console.WriteLine($"Код категории: {CategoryCode}");
    }

    public void ShowEmployeeLimit()
    {
        Console.WriteLine($"Максимум сотрудников: {MaxEmployees}");
    }

    public void ShowLoadingZone()
    {
        Console.WriteLine($"Зона погрузки: {HasLoadingZone}");
    }


    // Новые методы Sprint4

    public void ShowDepartmentCode()
    {
        Console.WriteLine($"Код отдела: {DepartmentCode}");
    }

    public void ShowShiftCount()
    {
        Console.WriteLine($"Количество смен: {ShiftCount}");
    }

    public void ShowForklift()
    {
        Console.WriteLine($"Наличие погрузчика: {HasForklift}");
    }
}


// -------------------------------------------------
// AutomatedInventory
// -------------------------------------------------

class AutomatedInventory : Inventory
{
    private string automationLevel;

    public string AutomationLevel
    {
        get { return automationLevel; }
        set { automationLevel = value; }
    }

    // Sprint2
    public int RobotCount { get; set; }
    public string SystemName { get; set; }
    public bool IsSystemActive { get; set; }

    // Sprint3
    public string SoftwareVersion { get; set; }
    public int SensorCount { get; set; }
    public bool BackupEnabled { get; set; }

    // Новые атрибуты Sprint4
    public string MaintenanceDate { get; set; }
    public double EnergyConsumption { get; set; }
    public int ErrorCount { get; set; }


    public AutomatedInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string automationLevel)
        : base(warehouseId, warehouseName, storageCapacity)
    {
        AutomationLevel = automationLevel;
    }


    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);

        Console.WriteLine(
            $"Уровень автоматизации: {AutomationLevel}"
        );
    }


    // Sprint2

    public void StartSystem()
    {
        IsSystemActive = true;
        Console.WriteLine($"Система {SystemName} запущена");
    }

    public void StopSystem()
    {
        IsSystemActive = false;
        Console.WriteLine($"Система {SystemName} остановлена");
    }

    public void ShowAutomationInfo()
    {
        Console.WriteLine($"Система: {SystemName}");
        Console.WriteLine($"Количество роботов: {RobotCount}");
        Console.WriteLine($"Уровень автоматизации: {AutomationLevel}");
    }


    // Sprint3

    public void ShowSoftwareVersion()
    {
        Console.WriteLine($"Версия программы: {SoftwareVersion}");
    }

    public void ShowSensorCount()
    {
        Console.WriteLine($"Количество датчиков: {SensorCount}");
    }

    public void ShowBackupStatus()
    {
        Console.WriteLine($"Резервное копирование: {BackupEnabled}");
    }


    // Новые методы Sprint4

    public void ShowMaintenanceDate()
    {
        Console.WriteLine($"Дата обслуживания: {MaintenanceDate}");
    }

    public void ShowEnergyConsumption()
    {
        Console.WriteLine($"Энергопотребление: {EnergyConsumption}");
    }

    public void ShowErrorCount()
    {
        Console.WriteLine($"Количество ошибок: {ErrorCount}");
    }
}


// -------------------------------------------------
// Сложное наследование из Sprint2
// -------------------------------------------------

class SmartAutomatedInventory : AutomatedInventory
{
    public string ControlSystem { get; set; }

    public SmartAutomatedInventory(
        int warehouseId,
        string warehouseName,
        double storageCapacity,
        string automationLevel,
        string controlSystem)
        : base(
            warehouseId,
            warehouseName,
            storageCapacity,
            automationLevel)
    {
        ControlSystem = controlSystem;
    }

    public void ShowControlSystem()
    {
        Console.WriteLine(
            $"Система управления: {ControlSystem}"
        );
    }
}


// -------------------------------------------------
// Generic-класс Sprint3
// -------------------------------------------------

class InventoryData<T>
{
    public T Data { get; set; }

    public InventoryData(T data)
    {
        Data = data;
    }

    public void ShowData()
    {
        Console.WriteLine($"Данные склада: {Data}");
    }
}


// -------------------------------------------------
// Управление зависимостью Sprint4
// -------------------------------------------------

class InventoryService
{
    private IInventoryOperations inventory;

    public InventoryService(IInventoryOperations inventory)
    {
        this.inventory = inventory;
    }

    public void Add(Item item)
    {
        Console.WriteLine("Сервис добавляет товар:");
        inventory.AddItem(item);
    }

    public void Remove(Item item)
    {
        Console.WriteLine("Сервис удаляет товар:");
        inventory.RemoveItem(item);
    }
}


// -------------------------------------------------
// Создание товаров
// -------------------------------------------------

Item item1 = new Item("Ноутбук");
Item item2 = new Item("Телефон");
Item item3 = new Item("Принтер");


// -------------------------------------------------
// Создание складов
// -------------------------------------------------

PersonalInventory personalInventory =
    new PersonalInventory(
        1,
        "Личный склад",
        100,
        "Юлия"
    );

personalInventory.Address = "Тюмень";
personalInventory.CurrentItemCount = 10;
personalInventory.IsOpen = true;

personalInventory.AccessCode = "1234";
personalInventory.RoomNumber = 15;
personalInventory.IsLocked = false;

personalInventory.ManagerName = "Юлия";
personalInventory.Temperature = 20;
personalInventory.SecurityLevel = "Средний";

personalInventory.LastAccessDate = "07.09.2026";
personalInventory.MaxPersonalItems = 50;
personalInventory.HasInsurance = true;

// Sprint4
personalInventory.LastInspectionDate = "01.10.2026";
personalInventory.MaxWeight = 2000;
personalInventory.HasFireSystem = true;

personalInventory.StorageType = "Личное хранение";
personalInventory.RentalMonths = 12;
personalInventory.PaymentStatus = true;


GroupInventory groupInventory =
    new GroupInventory(
        2,
        "Склад электроники",
        500,
        "Электроника"
    );

groupInventory.Address = "Тюмень, ул. Республики";
groupInventory.CurrentItemCount = 250;
groupInventory.IsOpen = true;

groupInventory.EmployeeCount = 8;
groupInventory.SectionName = "Отдел электроники";
groupInventory.ResponsiblePerson = "Иван Петров";

groupInventory.ManagerName = "Иван Петров";
groupInventory.Temperature = 18;
groupInventory.SecurityLevel = "Высокий";

groupInventory.CategoryCode = "EL-01";
groupInventory.MaxEmployees = 15;
groupInventory.HasLoadingZone = true;

// Sprint4
groupInventory.LastInspectionDate = "05.10.2026";
groupInventory.MaxWeight = 10000;
groupInventory.HasFireSystem = true;

groupInventory.DepartmentCode = "DEP-01";
groupInventory.ShiftCount = 2;
groupInventory.HasForklift = true;


AutomatedInventory automatedInventory =
    new AutomatedInventory(
        3,
        "Автоматизированный склад",
        1000,
        "Высокий"
    );

automatedInventory.Address = "Тюмень, Промышленная зона";
automatedInventory.CurrentItemCount = 600;
automatedInventory.IsOpen = true;

automatedInventory.RobotCount = 20;
automatedInventory.SystemName = "AutoStorage";
automatedInventory.IsSystemActive = false;

automatedInventory.ManagerName = "Анна Смирнова";
automatedInventory.Temperature = 17;
automatedInventory.SecurityLevel = "Максимальный";

automatedInventory.SoftwareVersion = "2.1";
automatedInventory.SensorCount = 40;
automatedInventory.BackupEnabled = true;

// Sprint4
automatedInventory.LastInspectionDate = "10.10.2026";
automatedInventory.MaxWeight = 25000;
automatedInventory.HasFireSystem = true;

automatedInventory.MaintenanceDate = "15.10.2026";
automatedInventory.EnergyConsumption = 150.5;
automatedInventory.ErrorCount = 2;


SmartAutomatedInventory smartInventory =
    new SmartAutomatedInventory(
        4,
        "Умный склад",
        1500,
        "Полный",
        "SmartControl"
    );

smartInventory.SystemName = "SmartStorage";


// -------------------------------------------------
// Проверка новых атрибутов и методов Sprint4
// -------------------------------------------------

Console.WriteLine("=== Персональный склад ===");

personalInventory.GetStorageStatus();
personalInventory.ShowInspectionDate();
personalInventory.ShowMaxWeight();
personalInventory.CheckFireSystem();
personalInventory.ShowStorageType();
personalInventory.ShowRentalMonths();
personalInventory.ShowPaymentStatus();


Console.WriteLine();

Console.WriteLine("=== Групповой склад ===");

groupInventory.GetStorageStatus();
groupInventory.ShowInspectionDate();
groupInventory.ShowDepartmentCode();
groupInventory.ShowShiftCount();
groupInventory.ShowForklift();


Console.WriteLine();

Console.WriteLine("=== Автоматизированный склад ===");

automatedInventory.GetStorageStatus();
automatedInventory.ShowInspectionDate();
automatedInventory.ShowMaintenanceDate();
automatedInventory.ShowEnergyConsumption();
automatedInventory.ShowErrorCount();


// -------------------------------------------------
// Явная реализация интерфейсов Sprint4
// -------------------------------------------------

Console.WriteLine();
Console.WriteLine("=== Явная реализация интерфейсов ===");

IReportable reportableInventory = groupInventory;
IReservable reservableInventory = groupInventory;

reportableInventory.CreateReport();
reservableInventory.ReservePlace();


// -------------------------------------------------
// Управление зависимостью Sprint4
// -------------------------------------------------

Console.WriteLine();
Console.WriteLine("=== Управление зависимостью ===");

InventoryService personalService =
    new InventoryService(personalInventory);

personalService.Add(item1);
personalService.Remove(item1);


Console.WriteLine();

InventoryService automatedService =
    new InventoryService(automatedInventory);

automatedService.Add(item3);
automatedService.Remove(item3);


// -------------------------------------------------
// Generic Sprint3
// -------------------------------------------------

Console.WriteLine();
Console.WriteLine("=== Generic-класс ===");

InventoryData<string> warehouseNameData =
    new InventoryData<string>(
        personalInventory.WarehouseName
    );

warehouseNameData.ShowData();


// -------------------------------------------------
// Сложное наследование
// -------------------------------------------------

Console.WriteLine();
Console.WriteLine("=== Сложное наследование ===");

smartInventory.GetStorageStatus();
smartInventory.StartSystem();
smartInventory.ShowControlSystem();
```
